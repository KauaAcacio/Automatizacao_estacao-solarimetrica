Programa que monitora se um dado novo foi adicionado

```python
#Entrada de variáveis
ultima_linha_analisada = 0
ultimo_registro_analisado = 0
quantidade_linhas = len(df)

#Verificando se houve alteração
while True:
    df = pd.read_table("WS600_Estação_Compacta_2019.txt", header = 1, sep=",")
    numero_linha = len(df) - 1
    ultimo_registro = df["RECORD"][numero_linha]
    if ultimo_registro>ultimo_registro_analisado:
        dados_mais_recentes = df.iloc[-1:, 0:13]
        ultimo_registro_analisado = ultimo_registro
        print(dados_mais_recentes)
        print("------------------"*10)
    time.sleep(10)
```
